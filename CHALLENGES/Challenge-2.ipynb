```python
from qiskit import QuantumCircuit, transpile
from qiskit.visualization import plot_histogram
from qiskit_aer import AerSimulator
import matplotlib.pyplot as plt
```

```python
qc_a = QuantumCircuit(1, 1)

qc_a.ry(0.5, 0)
qc_a.measure(0, 0)

simulator = AerSimulator()
compiled_a = transpile(qc_a, simulator)
result_a = simulator.run(compiled_a, shots=100).result()
counts_a = result_a.get_counts()

print("Circuit A:", counts_a)
display(plot_histogram(counts_a, title="Circuit A: 0 occurs more frequently"))
```

```python
qc_b = QuantumCircuit(1, 1)

qc_b.ry(2.5, 0)
qc_b.measure(0, 0)

compiled_b = transpile(qc_b, simulator)
result_b = simulator.run(compiled_b, shots=100).result()
counts_b = result_b.get_counts()

print("Circuit B:", counts_b)
display(plot_histogram(counts_b, title="Circuit B: 1 occurs more frequently"))
```

```python
"""
Explanation:

Circuit A uses an RY gate with an angle of 0.5 radians,
making outcome 0 more likely.

Circuit B uses an RY gate with an angle of 2.5 radians,
making outcome 1 more likely.

Both circuits are executed for 100 shots. The measurement
counts may vary slightly due to quantum randomness.
"""
```

```python
"""
Challenge Question:
How can quantum gates change the probability of obtaining
0 or 1 when a qubit is measured?

Answer:
Quantum gates change the quantum state and amplitudes of a qubit.
These amplitudes determine the probabilities of measuring 0 or 1.
The RY gate rotates the qubit by a chosen angle, allowing us to
control these probabilities and create a biased quantum coin.
"""
```

```python
print("Circuit A counts:", counts_a)
print("Total shots A:", sum(counts_a.values()))

print("\nCircuit B counts:", counts_b)
print("Total shots B:", sum(counts_b.values()))

assert sum(counts_a.values()) == 100, "Circuit A shot count incorrect"
assert sum(counts_b.values()) == 100, "Circuit B shot count incorrect"

print("\nBoth circuits passed the 100-shot verification!")
```
